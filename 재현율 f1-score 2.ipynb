```python
import mne
import numpy as np
import torch
import torch.nn as nn
import torch.optim as optim
from torch.utils.data import DataLoader, Dataset, random_split
from sklearn.metrics import recall_score, f1_score, classification_report
import matplotlib.pyplot as plt
import os

# ✅ 데이터 경로 및 참가자 목록
base_dir = r"C:\Users\owner\Desktop\dataset\derivatives"
subjects = [f"sub-{str(i).zfill(3)}" for i in range(1, 66)]  # sub-001 ~ sub-065

# ✅ AD/Control 라벨 기준
def get_label(subject_id):
    num = int(subject_id.split("-")[1])
    if num <= 36:
        return 1  # AD
    elif num <= 65:
        return 0  # Control
    else:
        return None

# ✅ EEG 데이터 로드 및 전처리
eeg_data = []
labels = []

print("🔄 EEG 데이터 로드 중...")
for subject in subjects:
    file_path = os.path.join(base_dir, subject, "eeg", f"{subject}_task-eyesclosed_eeg.set")
    if os.path.exists(file_path):
        print(f"✅ {subject} 데이터 로드")
        raw = mne.io.read_raw_eeglab(file_path, preload=True)
        raw.resample(128)
        raw.filter(1., 40.)
        data = raw.get_data()
        data = np.transpose(data, (1, 0))  # (time, channels)

        if data.shape[0] >= 1280:
            data = data[:1280, :]
        else:
            print(f"⚠ {subject}: 데이터 길이 부족, 스킵")
            continue

        label = get_label(subject)
        if label is not None:
            eeg_data.append(data)
            labels.append(label)
        else:
            print(f"❌ 라벨 정보 없음: {subject}, 스킵")
    else:
        print(f"❌ 파일 없음: {file_path}")

# ✅ 데이터 텐서화
if len(eeg_data) == 0:
    raise ValueError("❌ EEG 데이터를 불러오지 못했습니다.")

eeg_data_array = np.stack(eeg_data)  # (samples, time, channels)
labels_array = np.array(labels)
print("📊 데이터 shape:", eeg_data_array.shape, "라벨 shape:", labels_array.shape)

# -------------------------------
# ✅ PyTorch Dataset 정의
# -------------------------------
class EEGDataset(Dataset):
    def __init__(self, data, labels):
        self.data = torch.tensor(data, dtype=torch.float32)
        self.labels = torch.tensor(labels, dtype=torch.long)

    def __len__(self):
        return len(self.data)

    def __getitem__(self, idx):
        return self.data[idx], self.labels[idx]

# -------------------------------
# ✅ CNN + LSTM 모델 정의
# -------------------------------
class CNN_RNN_Model(nn.Module):
    def __init__(self):
        super(CNN_RNN_Model, self).__init__()
        self.conv1 = nn.Conv2d(1, 16, kernel_size=(3, 3), padding=1)
        self.pool = nn.MaxPool2d((2, 2))

        dummy_input = torch.zeros(1, 1, 1280, 19)
        dummy_output = self.pool(torch.relu(self.conv1(dummy_input)))
        time_steps = dummy_output.shape[2]
        feature_size = dummy_output.shape[1] * dummy_output.shape[3]
        self.lstm_input_size = feature_size
        print(f"✅ LSTM 입력 feature size: {self.lstm_input_size}, time_steps: {time_steps}")

        self.lstm = nn.LSTM(input_size=self.lstm_input_size, hidden_size=64, batch_first=True)
        self.fc1 = nn.Linear(64, 32)
        self.fc2 = nn.Linear(32, 2)
        self.softmax = nn.Softmax(dim=1)

    def forward(self, x):
        x = x.unsqueeze(1)  # (batch, 1, time, channels)
        x = self.pool(torch.relu(self.conv1(x)))
        x = x.permute(0, 2, 1, 3).contiguous()
        x = x.view(x.size(0), x.size(1), -1)
        x, _ = self.lstm(x)
        x = torch.relu(self.fc1(x[:, -1, :]))
        x = self.fc2(x)
        return self.softmax(x)

# -------------------------------
# ✅ 데이터 분할 및 DataLoader 생성
# -------------------------------
dataset = EEGDataset(eeg_data_array, labels_array)
train_size = int(0.8 * len(dataset))
val_size = len(dataset) - train_size
train_dataset, val_dataset = random_split(dataset, [train_size, val_size])

train_loader = DataLoader(train_dataset, batch_size=8, shuffle=True)
val_loader = DataLoader(val_dataset, batch_size=8)

# -------------------------------
# ✅ 모델 학습 설정 (Control에 적당한 가중치 적용)
# -------------------------------
model = CNN_RNN_Model()

# ✅ 실험 2: Control 클래스에 약간 가중치
class_weights = torch.tensor([1.5, 1.0], dtype=torch.float32)
criterion = nn.CrossEntropyLoss(weight=class_weights)

optimizer = optim.Adam(model.parameters(), lr=0.001)
epochs = 5
train_acc_history = []

print("🚀 딥러닝 모델 학습 시작...")
for epoch in range(epochs):
    model.train()
    running_loss, correct, total = 0.0, 0, 0
    for inputs, targets in train_loader:
        optimizer.zero_grad()
        outputs = model(inputs)
        loss = criterion(outputs, targets)
        loss.backward()
        optimizer.step()

        _, predicted = torch.max(outputs, 1)
        total += targets.size(0)
        correct += (predicted == targets).sum().item()
        running_loss += loss.item()

    accuracy = 100 * correct / total
    train_acc_history.append(accuracy)
    print(f"Epoch {epoch+1}, Loss: {running_loss:.3f}, Accuracy: {accuracy:.2f}%")

# -------------------------------
# ✅ 검증 및 평가지표 출력
# -------------------------------
model.eval()
all_preds = []
all_targets = []

with torch.no_grad():
    for inputs, targets in val_loader:
        outputs = model(inputs)
        probs = torch.softmax(outputs, dim=1)
        predicted = torch.argmax(probs, dim=1)

        all_preds.extend(predicted.cpu().numpy())
        all_targets.extend(targets.cpu().numpy())

# ✅ 평가지표 출력
accuracy = np.mean(np.array(all_preds) == np.array(all_targets)) * 100
recall = recall_score(all_targets, all_preds, pos_label=1)
f1 = f1_score(all_targets, all_preds, pos_label=1)

print(f"\n✅ 검증 정확도: {accuracy:.2f}%")
print(f"✅ 재현율 (Recall - AD 기준): {recall:.2f}")
print(f"✅ F1-Score (AD 기준): {f1:.2f}")
print("\n📋 Classification Report:")
print(classification_report(all_targets, all_preds, target_names=["Control", "AD"]))

# -------------------------------
# ✅ 학습 정확도 시각화
# -------------------------------
plt.plot(range(1, epochs+1), train_acc_history, label="Train Accuracy")
plt.xlabel("Epoch")
plt.ylabel("Accuracy (%)")
plt.title("CNN+RNN Training Accuracy (class weight [1.5, 1.0])")
plt.legend()
plt.show()
```
